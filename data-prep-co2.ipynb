```python
import os

healthy_path = "/kaggle/input/datasets/emmarex/plantdisease/PlantVillage/Tomato_healthy"
mosaic_path = "/kaggle/input/datasets/emmarex/plantdisease/PlantVillage/Tomato__Tomato_mosaic_virus"

healthy_images = [
    f for f in os.listdir(healthy_path)
    if f.lower().endswith((".jpg", ".jpeg", ".png"))
]

mosaic_images = [
    f for f in os.listdir(mosaic_path)
    if f.lower().endswith((".jpg", ".jpeg", ".png"))
]

print(f"Healthy images: {len(healthy_images)}")
print(f"Mosaic Virus images: {len(mosaic_images)}")
```

```python
import shutil
from pathlib import Path
from sklearn.model_selection import train_test_split

# ==========================================================
# Source Directories (Verified)
# ==========================================================
healthy_source = Path(
    "/kaggle/input/datasets/emmarex/plantdisease/PlantVillage/Tomato_healthy"
)

mosaic_source = Path(
    "/kaggle/input/datasets/emmarex/plantdisease/PlantVillage/Tomato__Tomato_mosaic_virus"
)

# ==========================================================
# Output Directory
# ==========================================================
output_dir = Path("/kaggle/working/dataset")

# ==========================================================
# Create Folder Structure
# ==========================================================
for split in ["train", "validation", "test"]:
    for cls in ["Healthy", "Mosaic"]:
        (output_dir / split / cls).mkdir(parents=True, exist_ok=True)

# ==========================================================
# Function to Split and Copy Images
# ==========================================================
def prepare_class(source_folder, class_name):

    # Collect image files only
    images = sorted([
        file for file in source_folder.iterdir()
        if file.suffix.lower() in [".jpg", ".jpeg", ".png"]
    ])

    # 70% Train, 30% Temp
    train_files, temp_files = train_test_split(
        images,
        test_size=0.30,
        random_state=42,
        shuffle=True
    )

    # Split remaining 30% into 15% Validation and 15% Test
    val_files, test_files = train_test_split(
        temp_files,
        test_size=0.50,
        random_state=42,
        shuffle=True
    )

    # Helper function
    def copy_files(files, split_name):
        destination = output_dir / split_name / class_name

        for file in files:
            shutil.copy2(file, destination / file.name)

    copy_files(train_files, "train")
    copy_files(val_files, "validation")
    copy_files(test_files, "test")

    print(f"\n{class_name}")
    print(f"Train      : {len(train_files)}")
    print(f"Validation : {len(val_files)}")
    print(f"Test       : {len(test_files)}")

# ==========================================================
# Prepare Both Classes
# ==========================================================
prepare_class(healthy_source, "Healthy")
prepare_class(mosaic_source, "Mosaic")

print("\n✅ Dataset preparation completed successfully.")
```

```python
# checking for corrupted images
from PIL import Image
from pathlib import Path

dataset_path = Path("/kaggle/working/dataset")

corrupted = []

for image_file in dataset_path.rglob("*"):
    if image_file.suffix.lower() in [".jpg", ".jpeg", ".png"]:
        try:
            with Image.open(image_file) as img:
                img.verify()
        except Exception:
            corrupted.append(str(image_file))

print("=" * 50)
print(f"Corrupted Images Found: {len(corrupted)}")

if corrupted:
    print("\nCorrupted Files:")
    for file in corrupted:
        print(file)
else:
    print("✅ No corrupted images found.")
```
